````python
# Gemini API setup
# In Google Colab, add your GEMINI_API_KEY to Colab Secrets before running this cell.

from google.colab import userdata
import google.generativeai as genai

api_key = userdata.get("GEMINI_API_KEY")
genai.configure(api_key=api_key)

model = genai.GenerativeModel("gemini-3.5-flash")

def get_response(prompt: str) -> str:
    response = model.generate_content(prompt)
    return response.text
````

## Task 1: The Explainer

**Prompt:** Explain the main difference between a Generative Adversarial Network (GAN) and a Variational Autoencoder (VAE).

````python
prompt_1 = "Explain the main difference between a Generative Adversarial Network (GAN) and a Variational Autoencoder (VAE)."
output_1 = get_response(prompt_1)
print(output_1)
````

### Model Output

A **Variational Autoencoder (VAE)** and a **Generative Adversarial Network (GAN)** are both generative AI models, but they learn to generate data in different ways.

- **VAE:** A VAE uses an **encoder and decoder**. The encoder maps input data into a structured probabilistic latent space, and the decoder generates data from that latent representation. VAEs are generally easier to control and provide a smooth, meaningful latent space, but their generated images can sometimes look blurry.
- **GAN:** A GAN uses two neural networks: a **generator** and a **discriminator**. The generator creates fake data, while the discriminator tries to distinguish real data from generated data. The two networks compete during training, which can produce very realistic and sharp outputs, although GANs can be harder to train.

**Main difference:** A VAE learns a probabilistic latent representation and reconstructs data, while a GAN learns through competition between a generator and a discriminator.
