```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ---- Step 1: Load and Explore the Data ----
data_path = '/content/sample_data/diabetes.csv'
if not os.path.exists(data_path):
    raise FileNotFoundError(
        f"Expected dataset not found at: {data_path}. Please place diabetes.csv at this location or update data_path to the correct file path."
    )
df = pd.read_csv(data_path)

print('Student: Supriya Misal')
print('\nFirst 5 rows:')
print(df.head())

print('\nDataset shape:', df.shape)
print('\nMissing values per column:')
print(df.isnull().sum())

# ---- Step 2: Prepare the Data ----
X = df.drop(columns=['Outcome'])
y = df['Outcome']

# Different split calculation from the original assignment: 25% test data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=7, stratify=y
)

print('\nTraining samples:', len(X_train))
print('Testing samples:', len(X_test))

# Scale the features before training
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ---- Step 3: Choose and Train a Model ----
# Different model setting: stronger regularization and a different solver
model = LogisticRegression(C=0.5, solver='liblinear', max_iter=1500)
model.fit(X_train_scaled, y_train)

# ---- Step 4: Evaluate the Model ----
y_pred = model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)

print(f'\nModel accuracy on test data: {accuracy:.2%}')
print('Correct predictions:', (y_test.to_numpy() == y_pred).sum())
print('Total test predictions:', len(y_test))
```
