This version follows the same NLP workflow as the reference assignment, with a different text corpus and additional word-frequency calculations.

```python
import nltk
import numpy as np
import pandas as pd
from collections import Counter
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

# One-time NLTK downloads
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

# Student: Supriya Misal
# Different sample corpus: technology/product reviews
corpus = [
    'The new laptop is fast, lightweight, and easy to use.',
    'The battery lasts long, but the keyboard feels uncomfortable.',
    'I enjoyed the bright display and excellent performance.'
]
text = ' '.join(corpus)

print('Student: Supriya Misal')
print('\nOriginal text:', text)

# ---------------------------------------------------------------
# 1. Tokenization: split text into sentences and words
# ---------------------------------------------------------------
sentences = sent_tokenize(text)
words = word_tokenize(text)
print('\nNumber of sentences:', len(sentences))
print('Sentences:', sentences)
print('Number of word tokens:', len(words))
print('Word tokens:', words)

# ---------------------------------------------------------------
# 2. Stopwords Removal: remove common low-information words
# ---------------------------------------------------------------
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in words
    if word.casefold() not in stop_words and word.isalpha()
]
print('\nFiltered words:', filtered_words)
print('Words after filtering:', len(filtered_words))

# ---------------------------------------------------------------
# 3. Stemming: reduce words to approximate root forms
# ---------------------------------------------------------------
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print('\nStemmed words:', stemmed_words)

# ---------------------------------------------------------------
# 4. Lemmatization: reduce words to dictionary forms
# ---------------------------------------------------------------
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in filtered_words]
print('\nLemmatized words:', lemmas)

# ---------------------------------------------------------------
# 5. Different calculation: word-frequency analysis
# ---------------------------------------------------------------
word_frequency = Counter(word.casefold() for word in filtered_words)
frequency_table = pd.DataFrame(
    word_frequency.items(), columns=['Word', 'Frequency']
).sort_values(['Frequency', 'Word'], ascending=[False, True])
print('\nWord frequency table:')
print(frequency_table.to_string(index=False))

# ---------------------------------------------------------------
# 6. Different calculation: Bag-of-Words representation
# ---------------------------------------------------------------
vectorizer = CountVectorizer(stop_words='english')
bow_matrix = vectorizer.fit_transform(corpus)
bow_df = pd.DataFrame(
    bow_matrix.toarray(), columns=vectorizer.get_feature_names_out()
)
print('\nBag-of-Words matrix:')
print(bow_df)
```
