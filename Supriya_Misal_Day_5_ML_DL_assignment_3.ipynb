MNIST neural-network classification with modified model settings and calculations.

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

# Student: Supriya Misal

# ---- Step 1: Load and Preprocess the Data ----
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize pixel values from [0, 255] to [0, 1]
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print('Student: Supriya Misal')
print('Training images:', x_train.shape[0])
print('Testing images:', x_test.shape[0])

# ---- Step 2: Build a Different Neural Network Model ----
# Modified calculations/settings: two hidden layers and dropout.
model = keras.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.Dropout(0.25),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

# ---- Step 3: Compile the Model ----
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0008),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ---- Step 4: Train and Evaluate the Model ----
# Different training calculation: 7 epochs, batch size 64, validation split 15%.
history = model.fit(
    x_train,
    y_train,
    epochs=7,
    batch_size=64,
    validation_split=0.15,
    verbose=1
)

test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f'\nFinal test loss: {test_loss:.4f}')
print(f'Final test accuracy: {test_accuracy:.2%}')

# ---- Step 5: Visualize Predictions ----
predictions = model.predict(x_test[:12], verbose=0)
predicted_labels = np.argmax(predictions, axis=1)

plt.figure(figsize=(12, 6))
for i in range(12):
    plt.subplot(3, 4, i + 1)
    plt.imshow(x_test[i], cmap='gray')
    plt.title(f'True: {y_test[i]} | Pred: {predicted_labels[i]}')
    plt.axis('off')

plt.suptitle('MNIST Predictions - Supriya Misal')
plt.tight_layout()
plt.show()
```
