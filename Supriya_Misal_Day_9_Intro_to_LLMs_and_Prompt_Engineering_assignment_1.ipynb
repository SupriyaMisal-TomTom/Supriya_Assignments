```python
import google.generativeai as genai
from google.colab import userdata
import nltk
from nltk.tokenize import word_tokenize
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Download NLTK data for tokenization
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Configure Gemini API
# Make sure GEMINI_API_KEY is saved in Colab's Secrets tab (key icon on left)
api_key = userdata.get("GEMINI_API_KEY")
genai.configure(api_key=api_key)

print("Setup completed successfully.")
```

```python
print("--- ASSIGNMENT 1: TASK 1 (Ticket Routing with Few-Shot) ---")

task1_prompt = """

Classify the following customer support emails into one of these departments: Shipping, Billing, or Returns.

Email: "Where is my package? The tracking link hasn't updated in 4 days."
Department: Shipping

Email: "I noticed a duplicate charge on my credit card statement for order #GT-10293."
Department: Billing

Email: "The sweater I bought is two sizes too small. I would like to send it back for a refund."
Department: Returns

Email: "Hi, I received my order (#GT-58291) today, but the headphones inside the box are broken. How do I send them back for a replacement?"
Department:

"""

model_default = genai.GenerativeModel("gemini-3.5-flash")
response_t1 = model_default.generate_content(task1_prompt)
print("Classification Output:\n", response_t1.text.strip())
```

```python

print("\n--- ASSIGNMENT 1: TASK 2 (Generating Responses with Temperature) ---")

# 1. Low Temperature (0.2) - Standard and Professional Apology
prompt_apology_pro = (
    "Write a standard, professional, and formal apology email to a customer whose order arrived late."
)
response_t2_pro = model_default.generate_content(
    prompt_apology_pro,
    generation_config=genai.types.GenerationConfig(temperature=0.2)
)
print("[Temperature 0.2 - Professional Apology]:\n", response_t2_pro.text.strip())

```

```python
# 2. High Temperature (0.9) - Empathetic and Creative Apology
prompt_apology_creative = (
    "Write an empathetic, warm, creative, and personalized apology email to a customer whose order arrived late."
)
response_t2_creative = model_default.generate_content(
    prompt_apology_creative,
    generation_config=genai.types.GenerationConfig(temperature=0.9)
)
print("\n[Temperature 0.9 - Creative Apology]:\n", response_t2_creative.text.strip())


print("\n--- ASSIGNMENT 1: TASK 3 (Answering Questions with Context) ---")

task3_prompt = """You are a helpful customer support assistant for Gadgetron.
Answer customer questions strictly using the official Return Policy provided below.
Do not assume or extrapolate information. If the answer cannot be determined directly from the policy, respond exactly with: "I do not have that information."

Official Return Policy:
"Customers can return any unused item in its original packaging within 30 days of purchase for a full refund. The customer is responsible for return shipping costs. Sale items are non-refundable."

Customer Question:
"Can I return a laptop I bought on sale last week? And will you pay for the shipping?"

Answer:"""

response_t3 = model_default.generate_content(task3_prompt)
print("Context-Based Output:\n", response_t3.text.strip())
```
