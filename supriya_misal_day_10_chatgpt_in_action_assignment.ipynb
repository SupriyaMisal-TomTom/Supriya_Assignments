This notebook uses OpenAI as primary engine and Gemini as fallback if OpenAI is unavailable.

```python
# =====================================================================
# Setup & Engine Health Check
# =====================================================================
!pip install -q openai==1.51.2 google-generativeai==0.8.3

from google.colab import userdata
import google.generativeai as genai
from openai import OpenAI

# 1. Fetch Keys from Colab Secrets
openai_key = userdata.get("OPENAI_API_KEY")
gemini_key = userdata.get("GEMINI_API_KEY")

openai_available = False
client_openai = None
model_gemini = None
gemini_model_name = "gemini-1.5-flash"

# 2. Check & Initialize OpenAI
if openai_key:
    try:
        temp_client = OpenAI(api_key=openai_key)
        # Lightweight test request to verify key/credits
        temp_client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": "ping"}],
            max_tokens=1,
        )
        client_openai = temp_client
        openai_available = True
    except Exception as e:
        print(f"[OpenAI Check]: Unavailable ({e}). Switching default to Gemini.")

# 3. Check & Initialize Gemini fallback
if gemini_key:
    try:
        genai.configure(api_key=gemini_key)
        model_gemini = genai.GenerativeModel(gemini_model_name)
    except Exception as e:
        print(f"[Gemini Setup Error]: {e}")


# 4. Unified calling function
def call_ai(prompt: str) -> tuple[str, str]:
    """Routes requests to OpenAI if active; otherwise to Gemini fallback."""
    global openai_available

    if openai_available and client_openai:
        try:
            response = client_openai.chat.completions.create(
                model="gpt-4o-mini",
                messages=[{"role": "user", "content": prompt}],
                temperature=0.7,
            )
            return response.choices[0].message.content.strip(), "OpenAI (gpt-4o-mini)"
        except Exception as e:
            print(f"[OpenAI Failed Mid-Run]: {e}\n--> Disabling OpenAI for remaining tasks...")
            openai_available = False

    if model_gemini:
        try:
            response = model_gemini.generate_content(prompt)
            return response.text.strip(), f"Google Gemini ({gemini_model_name})"
        except Exception as e:
            raise RuntimeError(f"Gemini generation error: {e}")

    raise ValueError(
        "No valid LLM provider available. Configure OPENAI_API_KEY (preferred) or GEMINI_API_KEY in Colab Secrets."
    )


# Display active engine status
if openai_available:
    print("Primary Engine: OpenAI (gpt-4o-mini)")
elif model_gemini:
    print(f"Primary Engine: Google Gemini ({gemini_model_name}) [Fallback Active]")
else:
    print("Warning: No working API keys detected.")
```

```python
# Task 1: The Data Extractor
task1_prompt = """Summarize the following text in a single sentence AND extract the key details into a valid JSON object.

Text:
"The new 'Starlight' laptop was released on October 26, 2024. It features a 14-inch OLED screen, weighs 2.5 lbs, and the base model with 16GB of RAM costs $1,299. The main complaint from ea[...]

Your response must follow this structure exactly:
Summary: <Single sentence summary>
JSON:
<Valid JSON object containing product_name, release_date, screen_size, screen_type, weight, ram, price, and main_complaint>"""

# Task 2: The Code Generator
task2_prompt = """Write a Python function named `calculate_average` that takes a list of numbers as input and returns the average of those numbers.

Requirements:
1. Handle the case where the list is empty by returning `None`.
2. Include a short docstring.
3. Provide only the Python code (no extra explanation)."""

# Execute Task 1 & Task 2
output_task1, model_task1 = call_ai(task1_prompt)
output_task2, model_task2 = call_ai(task2_prompt)
```

```python
# Task 3: The Code Translator (feeds on Task 2 output)
task3_prompt = f"""Using the exact Python code below, convert it into equivalent JavaScript (ES6+):

{output_task2}

Rules:
1. Preserve the same logic and behavior, including the empty-list handling.
2. Return only the JavaScript function with brief inline comments."""

# Task 4: The Social Media Manager
task4_prompt = """Act as a witty and enthusiastic social media manager. Write a tweet announcing the launch of the new 'Starlight' laptop.

Product Details:
- 14-inch OLED display
- Featherlight at 2.5 lbs
- 16GB RAM at $1,299

Strict constraints:
1. Include EXACTLY two emojis in total.
2. Include the hashtag #TechLaunch.
3. Return only the tweet text."""

# Execute Task 3 & Task 4
output_task3, model_task3 = call_ai(task3_prompt)
output_task4, model_task4 = call_ai(task4_prompt)
```

```python
tasks = [
    ("TASK 1: THE DATA EXTRACTOR", task1_prompt, output_task1, model_task1),
    ("TASK 2: THE CODE GENERATOR", task2_prompt, output_task2, model_task2),
    ("TASK 3: THE CODE TRANSLATOR", task3_prompt, output_task3, model_task3),
    ("TASK 4: THE SOCIAL MEDIA MANAGER", task4_prompt, output_task4, model_task4),
]

for title, prompt, output, used_model in tasks:
    print("=" * 90)
    print(f"{title} | Engine: {used_model}")
    print("=" * 90)
    print(f"EXACT PROMPT:\n{prompt}\n")
    print(f"AI FULL RESPONSE (UNEDITED):\n{output}\n")
```
