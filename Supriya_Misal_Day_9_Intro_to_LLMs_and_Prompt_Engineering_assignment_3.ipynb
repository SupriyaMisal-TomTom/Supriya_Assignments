```python
import google.generativeai as genai
from google.colab import userdata
import nltk
from nltk.tokenize import word_tokenize
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Download NLTK data for tokenization
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Configure Gemini API
# Make sure GEMINI_API_KEY is saved in Colab's Secrets tab (key icon on left)
api_key = userdata.get("GEMINI_API_KEY")
genai.configure(api_key=api_key)

print("Setup completed successfully.")
```

```python
model_default = genai.GenerativeModel("gemini-3.5-flash")
```

```python
print("--- ASSIGNMENT 3: TASK 1 (Tokenization) ---")

sentence = "Prompt engineering is the art of guiding AI models effectively."
tokens = word_tokenize(sentence)

print("Original Sentence:", sentence)
print("Tokens List:", tokens)
```

```python
print("\n--- ASSIGNMENT 3: TASK 2 (Semantic Similarity with Embeddings) ---")

sentences = [
    "The king is the ruler of the country.",
    "The queen governs her people with wisdom.",
    "I love a good pepperoni pizza.",
    "Monarchy is a form of government."
]

# Load embedding model
embed_model = SentenceTransformer('all-MiniLM-L6-v2')

# Generate embeddings
embeddings = embed_model.encode(sentences)

# Compute cosine similarity between sentence 0 and the rest (sentences 1, 2, 3)
base_embedding = embeddings[0].reshape(1, -1)
other_embeddings = embeddings[1:]

similarities = cosine_similarity(base_embedding, other_embeddings)[0]

# Display individual similarity scores
for idx, sim in enumerate(similarities, start=1):
    print(f"Similarity with '{sentences[idx]}': {sim:.4f}")

# Identify most similar sentence
best_match_idx = int(np.argmax(similarities) + 1)
print(f"\nMost semantically similar sentence to \"{sentences[0]}\":")
print(f"-> \"{sentences[best_match_idx]}\" (Cosine Similarity: {similarities[best_match_idx - 1]:.4f})")
```
