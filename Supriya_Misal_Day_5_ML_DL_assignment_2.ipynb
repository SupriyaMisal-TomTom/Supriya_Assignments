```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Student: Supriya Misal

# ---- Step 1: Load and Prepare the Data ----
df = pd.read_csv("/content/sample_data/Mall_Customers.csv")
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

print("Student: Supriya Misal")
print("\nFirst 5 customer records:")
print(df.head())

# ---- Step 2: Find the Optimal Number of Clusters (Elbow Method) ----
# Use a different range and random state from the original assignment.
wcss = []
cluster_range = range(2, 9)

for k in cluster_range:
    kmeans = KMeans(
        n_clusters=k,
        init="random",
        random_state=24,
        n_init=20
    )
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(9, 5))
plt.plot(list(cluster_range), wcss, marker="s")
plt.title("Elbow Method - Supriya Misal")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Within-Cluster Sum of Squares (WCSS)")
plt.xticks(list(cluster_range))
plt.grid(True, alpha=0.3)
plt.savefig("supriya_misal_elbow_plot.png")
plt.show()

# ---- Step 3: Train the K-Means Model ----
# Different calculation: use K=4 instead of K=5.
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    init="random",
    random_state=24,
    n_init=20
)

clusters = kmeans.fit_predict(X)
df["Cluster"] = clusters

print("\nNumber of customers in each cluster:")
print(df["Cluster"].value_counts().sort_index())

print("\nCluster centers:")
print(pd.DataFrame(
    kmeans.cluster_centers_,
    columns=["Annual Income (k$)", "Spending Score (1-100)"]
))

# ---- Step 4: Visualize the Clusters ----
plt.figure(figsize=(9, 6))
plt.scatter(
    X["Annual Income (k$)"],
    X["Spending Score (1-100)"],
    c=clusters,
    cmap="plasma",
    s=60
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c="black",
    marker="X",
    s=220,
    label="Cluster Centers"
)

plt.title("Customer Segments - Supriya Misal")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.legend()
plt.grid(True, alpha=0.2)
plt.savefig("supriya_misal_customer_segments.png")
plt.show()
```
